```python
# ============================================================
# SMART-PDM — LAYER 1
# STEP 2: RAW SIGNAL AUDIT  (نسخه اصلاح‌شده و کامل)
# ============================================================
#
# هدف:
#   بررسی کامل سیگنال‌های خام FAST و SLOW قبل از Modeling
#
# مهم:
#   - هیچ Windowing انجام نمی‌شود.
#   - هیچ Normalization انجام نمی‌شود.
#   - هیچ Model/Training انجام نمی‌شود.
#   - Split ثابت قبلی استفاده می‌شود.
#   - Unit of analysis = Cycle
#
# نکته اصلاحی مهم:
#   ساختار واقعی پوشه‌ها در این پروژه به شکل زیر است:
#       LAYER1_ROOT / split / cycle_id / fast.csv
#       LAYER1_ROOT / split / cycle_id / slow.csv
#   (بدون پوشه label در مسیر)
#
#   بنابراین از ستون "split_cycle_path" در Manifest استفاده می‌کنیم
#   و در صورت نبود، ساختارهای جایگزین را جستجو می‌کنیم.
#
# ============================================================


# ============================================================
# CELL 1 — IMPORTS
# ============================================================

from pathlib import Path
import json
import math
import warnings

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt

from IPython.display import display, Markdown

warnings.filterwarnings("ignore")

print("Imports loaded successfully.")


# ============================================================
# CELL 2 — CONFIGURATION
# ============================================================

RAW_ROOT = Path(r"D:\PROJECT\SMART-PDM-Dataset")
LAYER1_ROOT = Path(r"D:\PROJECT\SMART-PDM-Dataset-Layer1")

OUTPUT_ROOT = LAYER1_ROOT / "raw_signal_audit"
OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)

MANIFEST_PATH = LAYER1_ROOT / "layer1_fixed_split_manifest.csv"
CONFIG_PATH = LAYER1_ROOT / "layer1_split_config.json"

FAST_NAME = "fast.csv"
SLOW_NAME = "slow.csv"

RANDOM_SEED = 42

# FAST expected nominal sampling frequency (reference only).
EXPECTED_FAST_FS = 2048.0

# SLOW expected nominal sampling frequency.
EXPECTED_SLOW_FS = 1.0

# Tolerances used only for diagnostic classification.
FS_REL_TOL = 0.05

# Outlier threshold for robust audit:
# values outside Q1 - 3*IQR and Q3 + 3*IQR
OUTLIER_IQR_FACTOR = 3.0

# Maximum number of cycles used for plots.
MAX_PLOT_CYCLES_PER_SPLIT = 3

print("Configuration loaded.")


# ============================================================
# CELL 3 — LOAD AND VERIFY MANIFEST
# ============================================================

if not MANIFEST_PATH.exists():
    raise FileNotFoundError(
        f"Manifest not found:\n{MANIFEST_PATH}\n"
        "Run layer1_00_build_fixed_split first."
    )

manifest = pd.read_csv(MANIFEST_PATH)

print("Manifest shape:", manifest.shape)
print("\nColumns:")
print(manifest.columns.tolist())

display(manifest.head())


# ============================================================
# CELL 4 — VERIFY EXPECTED MANIFEST STRUCTURE
# ============================================================

required_manifest_columns = {
    "cycle_id",
    "label",
    "split"
}

missing_manifest_columns = required_manifest_columns - set(manifest.columns)

if missing_manifest_columns:
    raise ValueError(
        f"Manifest is missing required columns: "
        f"{sorted(missing_manifest_columns)}"
    )

print("Manifest structure: OK")

print("\nSplit distribution:")
display(
    manifest.groupby(["split", "label"])
    .size()
    .reset_index(name="cycles")
)

print("\nTotal cycles:", len(manifest))
print("Unique cycle IDs:", manifest["cycle_id"].nunique())


# ============================================================
# CELL 5 — LOAD SAVED CONFIG IF AVAILABLE
# ============================================================

saved_config = {}

if CONFIG_PATH.exists():
    with open(CONFIG_PATH, "r", encoding="utf-8") as f:
        saved_config = json.load(f)

    print("Saved Layer 1 configuration loaded.")

    print("\nSaved configuration:")
    print(json.dumps(saved_config, indent=2, ensure_ascii=False))
else:
    print("Warning: saved config not found.")


# ============================================================
# CELL 6 — BUILD COMPLETE CYCLE INVENTORY  (اصلاح‌شده)
# ============================================================

def find_cycle_file(cycle_id, split, label, filename, split_cycle_path=None):
    """
    Locate a specific signal file inside the fixed Layer 1 split.

    ترتیب جستجو:
      1) استفاده از split_cycle_path موجود در Manifest
      2) ساختار بدون label:  LAYER1_ROOT/split/cycle_id/filename
      3) ساختار با label:    LAYER1_ROOT/split/label/cycle_id/filename
      4) جستجوی بازگشتی (rglob) برای اطمینان
    """

    candidates = []

    # (1) مسیر ثبت‌شده در Manifest
    if split_cycle_path is not None:
        sp = str(split_cycle_path).strip()
        if sp not in ("", "None", "nan", "NaN"):
            candidates.append(Path(sp) / filename)

    # (2) ساختار بدون label
    candidates.append(LAYER1_ROOT / split / str(cycle_id) / filename)

    # (3) ساختار با label
    candidates.append(LAYER1_ROOT / split / label / str(cycle_id) / filename)

    # بررسی کاندیداها
    for c in candidates:
        if c.exists():
            return c

    # (4) جستجوی بازگشتی
    for base in [LAYER1_ROOT / split / label, LAYER1_ROOT / split]:
        if base.exists():
            for p in base.rglob(filename):
                if p.parent.name == str(cycle_id):
                    return p

    return None


# -------- تشخیص نام ستون مسیر در Manifest (اگر موجود باشد) --------
possible_path_columns = [
    "split_cycle_path",
    "cycle_path",
    "target_cycle_path",
    "output_cycle_path",
    "destination_cycle_path",
]

path_col = None
for c in possible_path_columns:
    if c in manifest.columns:
        path_col = c
        break

print(f"Using manifest path column: {path_col}")


inventory_rows = []

for _, row in manifest.iterrows():

    cycle_id = str(row["cycle_id"])
    split = str(row["split"])
    label = str(row["label"])

    split_cycle_path = row[path_col] if path_col else None

    fast_path = find_cycle_file(
        cycle_id, split, label, FAST_NAME, split_cycle_path
    )

    slow_path = find_cycle_file(
        cycle_id, split, label, SLOW_NAME, split_cycle_path
    )

    inventory_rows.append({
        "cycle_id": cycle_id,
        "split": split,
        "label": label,
        "fast_path": str(fast_path) if fast_path else None,
        "slow_path": str(slow_path) if slow_path else None,
        "fast_exists": fast_path is not None,
        "slow_exists": slow_path is not None
    })


inventory = pd.DataFrame(inventory_rows)

print("Inventory created.")
print("Rows:", len(inventory))

print("\nMissing FAST:")
print(int((~inventory["fast_exists"]).sum()))

print("Missing SLOW:")
print(int((~inventory["slow_exists"]).sum()))

if (~inventory["fast_exists"]).any() or (~inventory["slow_exists"]).any():
    display(
        inventory[
            (~inventory["fast_exists"]) |
            (~inventory["slow_exists"])
        ]
    )

    # --------- عیب‌یابی سریع قبل از پرتاب خطا ----------
    print("\n--- DIAGNOSTIC: exploring LAYER1_ROOT ---")
    for split_name in ["train", "validation", "test"]:
        p = LAYER1_ROOT / split_name
        print(f"\n[{split_name}] -> {p} exists={p.exists()}")
        if p.exists():
            subdirs = [x for x in p.iterdir() if x.is_dir()]
            for sd in subdirs[:5]:
                print("   subdir:", sd.name)
                for inner in list(sd.iterdir())[:5]:
                    print("      →", inner.name)

    sample_fast = list(LAYER1_ROOT.rglob("fast.csv"))[:5]
    print("\nSample fast.csv found on disk:")
    for s in sample_fast:
        print("   ", s)

    raise FileNotFoundError(
        "Some cycles are missing fast.csv or slow.csv. "
        "See diagnostics above."
    )

print("\nAll cycles contain both FAST and SLOW.")


# ============================================================
# CELL 7 — CSV READER
# ============================================================

def read_signal_csv(path):
    """
    Read CSV robustly.
    """
    try:
        df = pd.read_csv(path)
    except Exception:
        df = pd.read_csv(path, sep=None, engine="python")

    # Normalize column names
    df.columns = [str(c).strip() for c in df.columns]

    return df


# ============================================================
# CELL 8 — COLUMN TYPE DETECTION
# ============================================================

def detect_numeric_columns(df):
    """
    Return numeric columns.
    """
    numeric_cols = []

    for col in df.columns:
        converted = pd.to_numeric(df[col], errors="coerce")
        if np.isfinite(converted).sum() > 0:
            numeric_cols.append(col)

    return numeric_cols


def detect_time_columns(df):
    """
    Try to identify time/timestamp columns.
    """
    candidates = []

    for col in df.columns:
        name = str(col).lower()
        if any(
            key in name
            for key in ["time", "timestamp", "date", "datetime", "sample"]
        ):
            candidates.append(col)

    return candidates


def classify_signal_column(column_name):
    """
    Heuristic semantic classification.
    """
    name = str(column_name).lower()

    if any(k in name for k in ["vibration", "vib", "acceler", "accel"]):
        return "vibration_candidate"

    if any(k in name for k in ["current", "ampere", "amp", "irms", "i_rms"]):
        return "current_candidate"

    if any(
        k in name
        for k in ["active_power", "active power", "real_power", "real power", "power"]
    ):
        return "power_candidate"

    if "reactive" in name:
        return "reactive_power_candidate"

    if "apparent" in name:
        return "apparent_power_candidate"

    if any(k in name for k in ["voltage", "volt"]):
        return "voltage_candidate"

    if any(k in name for k in ["frequency", "freq"]):
        return "frequency_candidate"

    if any(k in name for k in ["power_factor", "power factor", "pf"]):
        return "power_factor_candidate"

    return "other_numeric"


# ============================================================
# CELL 9 — TIME / SAMPLING ANALYSIS
# ============================================================

def analyze_time_axis(df):
    """
    Detect possible time column and estimate sampling interval.
    """
    time_candidates = detect_time_columns(df)

    result = {
        "time_column": None,
        "time_type": None,
        "estimated_dt": np.nan,
        "estimated_fs": np.nan,
        "time_span": np.nan,
        "time_monotonic": None,
        "duplicate_timestamps": None
    }

    for col in time_candidates:

        s = df[col]

        # ---- Numeric time ----
        numeric = pd.to_numeric(s, errors="coerce")

        if numeric.notna().sum() >= max(3, int(0.8 * len(df))):

            values = numeric.dropna().to_numpy(dtype=float)

            if len(values) >= 3:

                diffs = np.diff(values)
                finite_diffs = diffs[np.isfinite(diffs)]
                positive_diffs = finite_diffs[finite_diffs > 0]

                if len(positive_diffs) > 0:

                    dt = np.median(positive_diffs)

                    if dt > 0:
                        result["time_column"] = col
                        result["time_type"] = "numeric"
                        result["estimated_dt"] = dt
                        result["estimated_fs"] = 1.0 / dt
                        result["time_span"] = values[-1] - values[0]
                        result["time_monotonic"] = bool(np.all(diffs >= 0))
                        result["duplicate_timestamps"] = bool(np.any(diffs == 0))
                        return result

        # ---- Datetime time ----
        parsed = pd.to_datetime(s, errors="coerce")

        if parsed.notna().sum() >= max(3, int(0.8 * len(df))):

            values = parsed.dropna()

            if len(values) >= 3:

                diffs = values.diff().dt.total_seconds().dropna().to_numpy()
                positive_diffs = diffs[diffs > 0]

                if len(positive_diffs) > 0:

                    dt = np.median(positive_diffs)

                    if dt > 0:
                        result["time_column"] = col
                        result["time_type"] = "datetime"
                        result["estimated_dt"] = dt
                        result["estimated_fs"] = 1.0 / dt
                        result["time_span"] = (
                            values.iloc[-1] - values.iloc[0]
                        ).total_seconds()
                        result["time_monotonic"] = bool(np.all(diffs >= 0))
                        result["duplicate_timestamps"] = bool(np.any(diffs == 0))
                        return result

    return result


# ============================================================
# CELL 10 — SIGNAL STATISTICS
# ============================================================

def calculate_signal_statistics(
    df, file_type, cycle_id, split, label, path
):

    numeric_cols = detect_numeric_columns(df)
    time_cols = detect_time_columns(df)

    rows = []

    for col in numeric_cols:

        values = pd.to_numeric(df[col], errors="coerce").to_numpy(dtype=float)

        total = len(values)

        finite_mask = np.isfinite(values)
        finite_values = values[finite_mask]

        nan_count = int(np.isnan(values).sum())
        inf_count = int(np.isinf(values).sum())
        missing_count = nan_count + inf_count

        if len(finite_values) == 0:

            rows.append({
                "cycle_id": cycle_id,
                "split": split,
                "label": label,
                "file_type": file_type,
                "file_path": str(path),
                "column": col,
                "column_role": classify_signal_column(col),
                "n_samples": total,
                "n_finite": 0,
                "nan_count": nan_count,
                "inf_count": inf_count,
                "missing_count": missing_count,
                "missing_ratio": 1.0,
                "min": np.nan,
                "max": np.nan,
                "mean": np.nan,
                "std": np.nan,
                "median": np.nan,
                "q01": np.nan,
                "q25": np.nan,
                "q75": np.nan,
                "q99": np.nan,
                "iqr": np.nan,
                "outlier_count": np.nan,
                "outlier_ratio": np.nan,
                "coefficient_variation": np.nan,
                "constant": True,
                "near_constant": True,
            })
            continue

        q01, q25, median, q75, q99 = np.percentile(
            finite_values, [1, 25, 50, 75, 99]
        )

        mean = float(np.mean(finite_values))
        std = float(np.std(finite_values, ddof=1)) if len(finite_values) > 1 else 0.0
        iqr = float(q75 - q25)

        lower = q25 - (OUTLIER_IQR_FACTOR * iqr)
        upper = q75 + (OUTLIER_IQR_FACTOR * iqr)

        outlier_mask = (finite_values < lower) | (finite_values > upper)
        outlier_count = int(outlier_mask.sum())
        outlier_ratio = outlier_count / len(finite_values)

        value_range = float(np.max(finite_values) - np.min(finite_values))

        scale = max(abs(mean), np.finfo(float).eps)
        coefficient_variation = abs(std) / scale

        constant = (value_range == 0)
        near_constant = (constant or coefficient_variation < 1e-6)

        rows.append({
            "cycle_id": cycle_id,
            "split": split,
            "label": label,
            "file_type": file_type,
            "file_path": str(path),
            "column": col,
            "column_role": classify_signal_column(col),
            "n_samples": total,
            "n_finite": len(finite_values),
            "nan_count": nan_count,
            "inf_count": inf_count,
            "missing_count": missing_count,
            "missing_ratio": missing_count / total,
            "min": float(np.min(finite_values)),
            "max": float(np.max(finite_values)),
            "mean": mean,
            "std": std,
            "median": float(median),
            "q01": float(q01),
            "q25": float(q25),
            "q75": float(q75),
            "q99": float(q99),
            "iqr": iqr,
            "outlier_count": outlier_count,
            "outlier_ratio": outlier_ratio,
            "coefficient_variation": coefficient_variation,
            "constant": constant,
            "near_constant": near_constant
        })

    return pd.DataFrame(rows)


# ============================================================
# CELL 11 — FILE-LEVEL AUDIT
# ============================================================

def audit_file(path, file_type, cycle_id, split, label):

    df = read_signal_csv(path)

    time_info = analyze_time_axis(df)

    numeric_cols = detect_numeric_columns(df)
    time_cols = detect_time_columns(df)

    total_cells = df.shape[0] * df.shape[1]
    total_missing = int(df.isna().sum().sum())

    numeric_missing = 0
    for col in numeric_cols:
        values = pd.to_numeric(df[col], errors="coerce")
        numeric_missing += int((~np.isfinite(values)).sum())

    duplicate_rows = int(df.duplicated().sum())

    missing_time_steps = np.nan
    nonpositive_time_steps = np.nan

    if time_info["time_column"] is not None:

        col = time_info["time_column"]

        numeric_time = pd.to_numeric(df[col], errors="coerce")

        if numeric_time.notna().sum() >= 3:
            t = numeric_time.dropna().to_numpy()
            diffs = np.diff(t)
        else:
            dt = pd.to_datetime(df[col], errors="coerce").dropna()
            diffs = dt.diff().dt.total_seconds().dropna().to_numpy()

        if len(diffs) > 0:

            positive = diffs[diffs > 0]

            if len(positive) > 0:
                median_dt = np.median(positive)
                missing_time_steps = int(
                    np.sum(np.abs(positive - median_dt) > 0.10 * median_dt)
                )

            nonpositive_time_steps = int(np.sum(diffs <= 0))

    result = {
        "cycle_id": cycle_id,
        "split": split,
        "label": label,
        "file_type": file_type,
        "file_path": str(path),
        "n_rows": len(df),
        "n_columns": len(df.columns),
        "columns": " | ".join(map(str, df.columns)),
        "numeric_columns": " | ".join(map(str, numeric_cols)),
        "time_columns": " | ".join(map(str, time_cols)),
        "n_numeric_columns": len(numeric_cols),
        "n_time_columns": len(time_cols),
        "time_column": time_info["time_column"],
        "time_type": time_info["time_type"],
        "estimated_dt": time_info["estimated_dt"],
        "estimated_fs": time_info["estimated_fs"],
        "time_span_seconds": time_info["time_span"],
        "time_monotonic": time_info["time_monotonic"],
        "duplicate_timestamps": time_info["duplicate_timestamps"],
        "missing_time_step_count": missing_time_steps,
        "nonpositive_time_step_count": nonpositive_time_steps,
        "total_missing_cells": total_missing,
        "numeric_invalid_values": numeric_missing,
        "duplicate_rows": duplicate_rows
    }

    return result, df


# ============================================================
# CELL 12 — RUN FILE-LEVEL AUDIT
# ============================================================

file_audit_rows = []
signal_stats_frames = []

total_files = len(inventory) * 2
counter = 0

for _, row in inventory.iterrows():

    cycle_id = row["cycle_id"]
    split = row["split"]
    label = row["label"]

    # -------- FAST --------
    fast_path = Path(row["fast_path"])

    fast_result, fast_df = audit_file(
        fast_path, "FAST", cycle_id, split, label
    )
    file_audit_rows.append(fast_result)

    fast_stats = calculate_signal_statistics(
        fast_df, "FAST", cycle_id, split, label, fast_path
    )
    signal_stats_frames.append(fast_stats)

    counter += 1
    print(f"[{counter}/{total_files}] Audited FAST: {cycle_id}")

    # -------- SLOW --------
    slow_path = Path(row["slow_path"])

    slow_result, slow_df = audit_file(
        slow_path, "SLOW", cycle_id, split, label
    )
    file_audit_rows.append(slow_result)

    slow_stats = calculate_signal_statistics(
        slow_df, "SLOW", cycle_id, split, label, slow_path
    )
    signal_stats_frames.append(slow_stats)

    counter += 1
    print(f"[{counter}/{total_files}] Audited SLOW: {cycle_id}")


file_audit = pd.DataFrame(file_audit_rows)
signal_stats = pd.concat(signal_stats_frames, ignore_index=True)

print("\nAudit completed.")
print("File-level rows:", len(file_audit))
print("Signal-stat rows:", len(signal_stats))


# ============================================================
# CELL 13 — SAVE RAW AUDIT TABLES
# ============================================================

file_audit_path = OUTPUT_ROOT / "file_level_audit.csv"
signal_stats_path = OUTPUT_ROOT / "signal_statistics.csv"
inventory_path = OUTPUT_ROOT / "audit_inventory.csv"

inventory.to_csv(inventory_path, index=False, encoding="utf-8-sig")
file_audit.to_csv(file_audit_path, index=False, encoding="utf-8-sig")
signal_stats.to_csv(signal_stats_path, index=False, encoding="utf-8-sig")

print("Saved:")
print(file_audit_path)
print(signal_stats_path)
print(inventory_path)


# ============================================================
# CELL 14 — COLUMN STRUCTURE AUDIT
# ============================================================

def summarize_columns(signal_stats, file_audit):

    rows = []

    for file_type in ["FAST", "SLOW"]:

        subset = signal_stats[signal_stats["file_type"] == file_type]

        for col in sorted(subset["column"].dropna().unique()):

            s = subset[subset["column"] == col]
            roles = sorted(s["column_role"].dropna().unique())

            rows.append({
                "file_type": file_type,
                "column": col,
                "role": " | ".join(roles),
                "cycles_with_column": s["cycle_id"].nunique(),
                "min_of_min": s["min"].min(),
                "max_of_max": s["max"].max(),
                "mean_of_means": s["mean"].mean(),
                "mean_std": s["std"].mean(),
                "max_missing_ratio": s["missing_ratio"].max(),
                "max_outlier_ratio": s["outlier_ratio"].max(),
                "constant_cycles": int(s["constant"].sum()),
                "near_constant_cycles": int(s["near_constant"].sum())
            })

    return pd.DataFrame(rows)


column_summary = summarize_columns(signal_stats, file_audit)

column_summary_path = OUTPUT_ROOT / "column_summary.csv"
column_summary.to_csv(column_summary_path, index=False, encoding="utf-8-sig")

print("COLUMN SUMMARY")
display(column_summary)


# ============================================================
# CELL 15 — FAST STRUCTURE
# ============================================================

fast_files = file_audit[file_audit["file_type"] == "FAST"].copy()

print("========== FAST STRUCTURE ==========")
print("\nNumber of FAST files:", len(fast_files))
print("Unique row counts:", sorted(fast_files["n_rows"].dropna().unique())[:30])
print("\nUnique column counts:", sorted(fast_files["n_columns"].dropna().unique()))

print("\nDetected columns:")
fast_columns = column_summary[column_summary["file_type"] == "FAST"][
    ["column", "role", "cycles_with_column"]
]
display(fast_columns)


# ============================================================
# CELL 16 — SLOW STRUCTURE
# ============================================================

slow_files = file_audit[file_audit["file_type"] == "SLOW"].copy()

print("========== SLOW STRUCTURE ==========")
print("\nNumber of SLOW files:", len(slow_files))
print("Unique row counts:", sorted(slow_files["n_rows"].dropna().unique())[:30])
print("\nUnique column counts:", sorted(slow_files["n_columns"].dropna().unique()))

print("\nDetected columns:")
slow_columns = column_summary[column_summary["file_type"] == "SLOW"][
    ["column", "role", "cycles_with_column"]
]
display(slow_columns)


# ============================================================
# CELL 17 — SAMPLING RATE AUDIT
# ============================================================

def sampling_summary(file_audit, file_type):

    subset = file_audit[file_audit["file_type"] == file_type].copy()
    summary_rows = []

    for split in ["train", "validation", "test"]:

        s = subset[subset["split"] == split]
        fs = s["estimated_fs"].dropna()

        if len(fs) == 0:
            summary_rows.append({
                "file_type": file_type,
                "split": split,
                "n_files": len(s),
                "fs_min": np.nan,
                "fs_median": np.nan,
                "fs_mean": np.nan,
                "fs_max": np.nan,
                "expected_fs": (
                    EXPECTED_FAST_FS if file_type == "FAST" else EXPECTED_SLOW_FS
                )
            })
        else:
            summary_rows.append({
                "file_type": file_type,
                "split": split,
                "n_files": len(s),
                "fs_min": fs.min(),
                "fs_median": fs.median(),
                "fs_mean": fs.mean(),
                "fs_max": fs.max(),
                "expected_fs": (
                    EXPECTED_FAST_FS if file_type == "FAST" else EXPECTED_SLOW_FS
                )
            })

    return pd.DataFrame(summary_rows)


fs_summary = pd.concat(
    [
        sampling_summary(file_audit, "FAST"),
        sampling_summary(file_audit, "SLOW")
    ],
    ignore_index=True
)

print("========== SAMPLING RATE SUMMARY ==========")
display(fs_summary)

fs_summary.to_csv(
    OUTPUT_ROOT / "sampling_rate_summary.csv",
    index=False,
    encoding="utf-8-sig"
)


# ============================================================
# CELL 18 — SAMPLING RATE CONSISTENCY
# ============================================================

def fs_status(fs, expected):

    if not np.isfinite(fs):
        return "UNDETERMINED"

    relative_error = abs(fs - expected) / expected

    if relative_error <= FS_REL_TOL:
        return "OK"

    return "CHECK"


fs_status_rows = []

for _, row in file_audit.iterrows():
    expected = (
        EXPECTED_FAST_FS if row["file_type"] == "FAST" else EXPECTED_SLOW_FS
    )
    fs_status_rows.append(fs_status(row["estimated_fs"], expected))

file_audit["sampling_rate_status"] = fs_status_rows

print(file_audit[["file_type", "sampling_rate_status"]].value_counts())


# ============================================================
# CELL 19 — SAMPLE COUNT / CYCLE DURATION
# ============================================================

duration_summary = (
    file_audit
    .groupby(["file_type", "split"])
    .agg(
        files=("cycle_id", "count"),
        min_samples=("n_rows", "min"),
        median_samples=("n_rows", "median"),
        mean_samples=("n_rows", "mean"),
        max_samples=("n_rows", "max"),
        min_duration_s=("time_span_seconds", "min"),
        median_duration_s=("time_span_seconds", "median"),
        mean_duration_s=("time_span_seconds", "mean"),
        max_duration_s=("time_span_seconds", "max")
    )
    .reset_index()
)

print("========== SAMPLE / DURATION SUMMARY ==========")
display(duration_summary)

duration_summary.to_csv(
    OUTPUT_ROOT / "sample_duration_summary.csv",
    index=False,
    encoding="utf-8-sig"
)


# ============================================================
# CELL 20 — DATA QUALITY: NaN / INF
# ============================================================

quality_summary = (
    file_audit
    .groupby(["file_type", "split"])
    .agg(
        files=("cycle_id", "count"),
        files_with_missing=("total_missing_cells", lambda x: int((x > 0).sum())),
        max_missing_cells=("total_missing_cells", "max"),
        max_numeric_invalid=("numeric_invalid_values", "max"),
        files_with_duplicate_rows=("duplicate_rows", lambda x: int((x > 0).sum()))
    )
    .reset_index()
)

print("========== DATA QUALITY ==========")
display(quality_summary)


# ============================================================
# CELL 21 — MISSING SAMPLE / TIME GAP AUDIT
# ============================================================

gap_summary = (
    file_audit
    .groupby(["file_type", "split"])
    .agg(
        files=("cycle_id", "count"),
        files_with_time_gap=(
            "missing_time_step_count",
            lambda x: int((x.fillna(0) > 0).sum())
        ),
        max_time_gap_count=("missing_time_step_count", "max"),
        files_with_nonpositive_time=(
            "nonpositive_time_step_count",
            lambda x: int((x.fillna(0) > 0).sum())
        )
    )
    .reset_index()
)

print("========== TIME / SAMPLE GAP AUDIT ==========")
display(gap_summary)


# ============================================================
# CELL 22 — CONSTANT / NEAR-CONSTANT CHANNEL AUDIT
# ============================================================

constant_summary = (
    signal_stats
    .groupby(["file_type", "column"])
    .agg(
        cycles=("cycle_id", "nunique"),
        constant_cycles=("constant", "sum"),
        near_constant_cycles=("near_constant", "sum"),
        max_missing_ratio=("missing_ratio", "max"),
        max_outlier_ratio=("outlier_ratio", "max")
    )
    .reset_index()
)

constant_summary["constant_ratio"] = (
    constant_summary["constant_cycles"] / constant_summary["cycles"]
)
constant_summary["near_constant_ratio"] = (
    constant_summary["near_constant_cycles"] / constant_summary["cycles"]
)

print("========== CONSTANT / NEAR-CONSTANT ==========")
display(constant_summary)


# ============================================================
# CELL 23 — OUTLIER AUDIT
# ============================================================

outlier_summary = (
    signal_stats
    .groupby(["file_type", "column"])
    .agg(
        cycles=("cycle_id", "nunique"),
        mean_outlier_ratio=("outlier_ratio", "mean"),
        median_outlier_ratio=("outlier_ratio", "median"),
        max_outlier_ratio=("outlier_ratio", "max"),
        total_outlier_count=("outlier_count", "sum")
    )
    .reset_index()
)

print("========== OUTLIER AUDIT ==========")
display(outlier_summary)


# ============================================================
# CELL 24 — SIGNAL RANGE SUMMARY
# ============================================================

range_summary = (
    signal_stats
    .groupby(["file_type", "column"])
    .agg(
        cycles=("cycle_id", "nunique"),
        global_min=("min", "min"),
        global_max=("max", "max"),
        mean_of_means=("mean", "mean"),
        mean_std=("std", "mean"),
        min_median=("median", "min"),
        max_median=("median", "max"),
        mean_q01=("q01", "mean"),
        mean_q99=("q99", "mean")
    )
    .reset_index()
)

print("========== SIGNAL RANGE SUMMARY ==========")
display(range_summary)


# ============================================================
# CELL 25 — MODALITY CANDIDATE DETECTION
# ============================================================

modality_candidates = {}

for file_type in ["FAST", "SLOW"]:

    subset = column_summary[column_summary["file_type"] == file_type]
    modality_candidates[file_type] = {}

    for role in [
        "vibration_candidate",
        "current_candidate",
        "power_candidate",
        "reactive_power_candidate",
        "apparent_power_candidate",
        "voltage_candidate",
        "frequency_candidate",
        "power_factor_candidate"
    ]:
        cols = subset[subset["role"] == role]["column"].tolist()
        modality_candidates[file_type][role] = cols


print("========== MODALITY CANDIDATES ==========")
print(json.dumps(modality_candidates, indent=2, ensure_ascii=False))


# ============================================================
# CELL 26 — SAVE MODALITY CANDIDATES
# ============================================================

with open(
    OUTPUT_ROOT / "modality_candidates.json", "w", encoding="utf-8"
) as f:
    json.dump(modality_candidates, f, indent=2, ensure_ascii=False)


# ============================================================
# CELL 27 — READ ONE REPRESENTATIVE FILE
# ============================================================

representative_cycle = (
    inventory
    .sort_values(["split", "label", "cycle_id"])
    .iloc[0]
)

print("Representative cycle:")
display(pd.DataFrame([representative_cycle]))

rep_fast = read_signal_csv(Path(representative_cycle["fast_path"]))
rep_slow = read_signal_csv(Path(representative_cycle["slow_path"]))

print("\nRepresentative FAST:")
print(rep_fast.shape)
display(rep_fast.head())

print("\nRepresentative SLOW:")
print(rep_slow.shape)
display(rep_slow.head())


# ============================================================
# CELL 28 — PLOT REPRESENTATIVE FAST SIGNALS
# ============================================================

numeric_fast_cols = detect_numeric_columns(rep_fast)

if len(numeric_fast_cols) == 0:
    print("No numeric FAST columns available for plotting.")
else:
    for col in numeric_fast_cols[:5]:
        values = pd.to_numeric(rep_fast[col], errors="coerce").to_numpy(dtype=float)

        plt.figure(figsize=(14, 4))
        plt.plot(values)
        plt.title(f"Representative FAST signal — {col}")
        plt.xlabel("Sample")
        plt.ylabel(col)
        plt.grid(alpha=0.3)
        plt.tight_layout()
        plt.show()


# ============================================================
# CELL 29 — PLOT REPRESENTATIVE SLOW SIGNALS
# ============================================================

numeric_slow_cols = detect_numeric_columns(rep_slow)

if len(numeric_slow_cols) == 0:
    print("No numeric SLOW columns available for plotting.")
else:
    for col in numeric_slow_cols:
        values = pd.to_numeric(rep_slow[col], errors="coerce").to_numpy(dtype=float)

        plt.figure(figsize=(14, 4))
        plt.plot(values, marker=".")
        plt.title(f"Representative SLOW signal — {col}")
        plt.xlabel("Sample")
        plt.ylabel(col)
        plt.grid(alpha=0.3)
        plt.tight_layout()
        plt.show()


# ============================================================
# CELL 30 — PLOT SAMPLE COUNTS
# ============================================================

for file_type in ["FAST", "SLOW"]:

    subset = file_audit[file_audit["file_type"] == file_type]

    plt.figure(figsize=(14, 5))
    for split in ["train", "validation", "test"]:
        values = subset[subset["split"] == split]["n_rows"]
        plt.hist(values, bins=20, alpha=0.5, label=split)

    plt.title(f"{file_type} sample-count distribution")
    plt.xlabel("Number of samples per file")
    plt.ylabel("Number of files")
    plt.legend()
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()


# ============================================================
# CELL 31 — PLOT SAMPLING RATE DISTRIBUTION
# ============================================================

for file_type in ["FAST", "SLOW"]:

    subset = file_audit[file_audit["file_type"] == file_type]
    fs_values = subset["estimated_fs"].dropna()

    if len(fs_values) == 0:
        continue

    plt.figure(figsize=(12, 4))
    plt.hist(fs_values, bins=20)

    expected = EXPECTED_FAST_FS if file_type == "FAST" else EXPECTED_SLOW_FS
    plt.axvline(expected, linestyle="--", label=f"Expected ≈ {expected} Hz")

    plt.title(f"{file_type} estimated sampling rate")
    plt.xlabel("Sampling frequency [Hz]")
    plt.ylabel("Number of files")
    plt.legend()
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()


# ============================================================
# CELL 32 — LABEL-WISE SIGNAL STATISTICS
# ============================================================

label_signal_summary = (
    signal_stats
    .groupby(["file_type", "label", "column"])
    .agg(
        cycles=("cycle_id", "nunique"),
        mean_mean=("mean", "mean"),
        mean_std=("std", "mean"),
        global_min=("min", "min"),
        global_max=("max", "max"),
        mean_outlier_ratio=("outlier_ratio", "mean"),
        max_missing_ratio=("missing_ratio", "max")
    )
    .reset_index()
)

label_signal_summary.to_csv(
    OUTPUT_ROOT / "label_signal_summary.csv",
    index=False,
    encoding="utf-8-sig"
)

print("========== LABEL-WISE SIGNAL SUMMARY ==========")
display(label_signal_summary.head(100))


# ============================================================
# CELL 33 — CYCLE DURATION DISTRIBUTION BY LABEL
# ============================================================

for file_type in ["FAST", "SLOW"]:

    subset = file_audit[file_audit["file_type"] == file_type]

    plt.figure(figsize=(14, 5))
    for label in sorted(subset["label"].unique()):
        values = subset[subset["label"] == label]["n_rows"]
        plt.hist(values, bins=20, alpha=0.5, label=label)

    plt.title(f"{file_type} sample count by class")
    plt.xlabel("Number of samples")
    plt.ylabel("Number of files")
    plt.legend()
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()


# ============================================================
# CELL 34 — FINAL AUTOMATED AUDIT REPORT
# ============================================================

report = {
    "total_cycles": int(len(manifest)),
    "total_files": int(len(file_audit)),
    "fast_files": int((file_audit["file_type"] == "FAST").sum()),
    "slow_files": int((file_audit["file_type"] == "SLOW").sum()),
    "missing_fast_files": int((~inventory["fast_exists"]).sum()),
    "missing_slow_files": int((~inventory["slow_exists"]).sum()),
    "files_with_missing_cells": int(
        (file_audit["total_missing_cells"] > 0).sum()
    ),
    "files_with_numeric_invalid_values": int(
        (file_audit["numeric_invalid_values"] > 0).sum()
    ),
    "files_with_duplicate_rows": int(
        (file_audit["duplicate_rows"] > 0).sum()
    ),
    "files_with_sampling_rate_check": int(
        (file_audit["sampling_rate_status"] == "CHECK").sum()
    ),
    "files_with_time_gaps": int(
        (file_audit["missing_time_step_count"].fillna(0) > 0).sum()
    ),
    "files_with_nonpositive_time_steps": int(
        (file_audit["nonpositive_time_step_count"].fillna(0) > 0).sum()
    )
}

critical_problems = []

if report["missing_fast_files"] > 0:
    critical_problems.append("Missing FAST files")

if report["missing_slow_files"] > 0:
    critical_problems.append("Missing SLOW files")

if report["files_with_numeric_invalid_values"] > 0:
    critical_problems.append("Numeric NaN/Inf detected")

if report["files_with_sampling_rate_check"] > 0:
    critical_problems.append("Sampling-rate inconsistency detected")

if report["files_with_time_gaps"] > 0:
    critical_problems.append("Possible time/sample gaps detected")

if len(critical_problems) == 0:
    report["overall_status"] = "PASS_WITH_REVIEW"
else:
    report["overall_status"] = "REVIEW_REQUIRED"

report["critical_problems"] = critical_problems

report_path = OUTPUT_ROOT / "raw_signal_audit_report.json"
with open(report_path, "w", encoding="utf-8") as f:
    json.dump(report, f, indent=2, ensure_ascii=False)


# ============================================================
# CELL 35 — HUMAN-READABLE FINAL REPORT
# ============================================================

print("\n")
print("=" * 70)
print("SMART-PDM — LAYER 1 RAW SIGNAL AUDIT")
print("=" * 70)

print(f"\nTotal cycles: {report['total_cycles']}")
print(f"Total signal files: {report['total_files']}")
print(f"FAST files: {report['fast_files']}")
print(f"SLOW files: {report['slow_files']}")

print(f"\nMissing FAST: {report['missing_fast_files']}")
print(f"Missing SLOW: {report['missing_slow_files']}")

print(f"\nFiles with NaN/Inf: {report['files_with_numeric_invalid_values']}")
print(f"Files with duplicate rows: {report['files_with_duplicate_rows']}")
print(f"Files with sampling-rate CHECK: {report['files_with_sampling_rate_check']}")
print(f"Files with possible time gaps: {report['files_with_time_gaps']}")
print(f"Files with non-positive time steps: {report['files_with_nonpositive_time_steps']}")

print("\nOverall status:", report["overall_status"])

if critical_problems:
    print("\nIssues requiring review:")
    for issue in critical_problems:
        print("  -", issue)
else:
    print("\nNo automatic critical issue detected.")

print("\nDetected FAST modalities:")
for role, cols in modality_candidates["FAST"].items():
    if cols:
        print(f"  {role}: {cols}")

print("\nDetected SLOW modalities:")
for role, cols in modality_candidates["SLOW"].items():
    if cols:
        print(f"  {role}: {cols}")

print("\nAudit outputs saved to:")
print(OUTPUT_ROOT)

print("\n" + "=" * 70)
print("RAW SIGNAL AUDIT FINISHED")
print("=" * 70)


# ============================================================
# CELL 36 — FINAL DECISION TABLE
# ============================================================

decision_table = pd.DataFrame({

    "Audit Item": [
        "Cycle-level split already fixed",
        "Every cycle has FAST",
        "Every cycle has SLOW",
        "FAST structure identified",
        "SLOW structure identified",
        "Sampling rate checked",
        "NaN/Inf checked",
        "Missing/time gaps checked",
        "Constant channels checked",
        "Outliers checked",
        "Modality candidates identified",
        "Windowing performed",
        "Modeling performed"
    ],

    "Status": [
        "YES",
        "YES" if report["missing_fast_files"] == 0 else "NO",
        "YES" if report["missing_slow_files"] == 0 else "NO",
        "YES",
        "YES",
        "YES",
        "YES",
        "YES",
        "YES",
        "YES",
        "YES",
        "NO — intentional",
        "NO — intentional"
    ]
})

display(decision_table)


# ============================================================
# END OF NOTEBOOK
# ============================================================
```
